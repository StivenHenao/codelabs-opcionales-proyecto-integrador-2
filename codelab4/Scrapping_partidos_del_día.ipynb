```python
# Instalar dependencias (ejecutar solo si no están instaladas)
# !pip install requests beautifulsoup4 lxml pandas

# Importar librerías
import requests
from bs4 import BeautifulSoup
import pandas as pd
import csv
import time
from datetime import datetime
import re

print("✅ Librerías importadas correctamente")
```

```python
# URL del sitio de partidos
url = "https://www.futbolred.com/parrilla-de-futbol"

# Headers para simular un navegador real
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
}

# Hacer la petición
response = requests.get(url, headers=headers)
print(f"Status Code: {response.status_code}")
print(f"Content Length: {len(response.text)}")

# Verificar que la petición fue exitosa
if response.status_code == 200:
    print("✅ Petición exitosa")
else:
    print("❌ Error en la petición")
```

```python
# soup ya contiene el HTML
soup = BeautifulSoup(response.text, 'html.parser')

# Buscar la tabla de partidos
table = soup.find('table', class_='tabla-dinamica')
partidos = []

if table:
    rows = table.find('tbody').find_all('tr')
    for row in rows:
        equipo = row.find('td', class_='partido').get_text(separator=" ", strip=True)
        liga = row.find('td', class_='liga').get_text(strip=True)
        hora = row.find('td', class_='hora').get_text(strip=True)
        canal = row.find('td', class_='canal').get_text(strip=True)
        partidos.append({
            "equipo": equipo,
            "liga": liga,
            "hora": hora,
            "canal": canal
        })

# Mostrar resultados
for p in partidos:
    print(p)
```

```python
import datetime

now = datetime.datetime.now()
fecha_str = now.strftime("%Y-%m-%d")

# Mostrar los partidos
print("Partidos del día: " + fecha_str + " Partido | Liga | Hora | Canal(es)")
for i, match in enumerate(partidos, 1):
    print(f"{i}. {match['equipo']} | {match['liga']} | {match['hora']} | {match['canal']}")
```
